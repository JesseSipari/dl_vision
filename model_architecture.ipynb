```python
from keras.models import Sequential
from keras.layers import Conv2D, MaxPooling2D, Flatten, Dense, Dropout
```

## Build the model object

```python
model = Sequential()
```

## CONV_1 adds a convolutional layer with ReLU activation and depth = 32 kernels

```python
model.add(Conv2D(32, kernel_size=(3, 3), strides=1, padding='same', activation='relu', input_shape=(28,28,1)))
```

## POOL_1: downsamples the image to choose the best features

```python
model.add(MaxPooling2D(pool_size=(2, 2)))
```

## CONV_2 increases the depth to 64

```python
model.add(Conv2D(64, (3, 3), strides=1, padding='same', activation='relu'))
```

## POOL_2: more downsampling

```python
model.add(MaxPooling2D(pool_size=(2, 2)))
```

## Flatten, since there are too many dimensions; we only want a calssification output

```python
model.add(Flatten())
```

## Add Dropout layer with 30% Probability

```python
model.add(Dropout(rate=0.3))
```

## FC_1: Fully connected to get all relevant data

```python
model.add(Dense(64, activation='relu'))
```

## Add dropout layer with 50% probability

```python
model.add(Dropout(rate=0.5))
```

## FC_2: Outputs a softmax to squash the matrix into output probabilities for the 10 classes

```python
model.add(Dense(10, activation='softmax'))
```

## Prints the model architecture summary

```python
model.summary()
```
